### Жадный алгоритм (найти максимальное количество взаимно непересекающихся интервалов)

```python
n = int(input())
otrezki = []
tochki = []

# Задаются исходные данные
for i in range(n):
    otrezok = list(map(int, input().split()))
    otrezki.append(otrezok)
    otrezki.sort(key = lambda x: x[1])
#     print(otrezki)
# В список правых точек добавляется первая точка
tochki = [otrezki[0][1]]
# Для каждого отрезка проверяется попадания в интервал
# и точка доавляется в список
for otr1 in otrezki:
    if otr1[0] > tochki[-1]:
        tochki.append(otr1[1])
print(len(tochki))
```

### Ханойские башни

```python
def tower_of_hanoi(numbers, start, auxiliary, end):  
    if numbers == 1:  
        print(start, end)  
        return
    tower_of_hanoi(numbers - 1, start, end, auxiliary)  
    print(start, end)  
    tower_of_hanoi(numbers - 1, auxiliary, start, end)  
n = int(input())  
print((2 ** n) - 1) 
tower_of_hanoi(n, 1, 2, 3)
```

### Сортировка выбором

```python
n = int(input())
list_for_sort = list(map(int, input().split()))
for i in range(len(list_for_sort) - 1):
    min_ = i
    for k in range(i + 1, len(list_for_sort)):
            if list_for_sort[k] < list_for_sort[min_]:
                min_ = k
    list_for_sort[i], list_for_sort[min_] = list_for_sort[min_], list_for_sort[i]
print(' '.join(str(el) for el in list_for_sort))
```

### Быстрая сортировка

```python
from random import randint
def quicksort(array):
    if len(array) < 2:
        return array
    low, same, high = [], [], []
    pivot = array[randint(0, len(array) - 1)]
    for item in array:
        if item < pivot:
            low.append(item)
        elif item == pivot:
            same.append(item)
        elif item > pivot:
            high.append(item)
    return quicksort(low) + same + quicksort(high)
n = int(input())
alist = list(map(int, input().split()))
list_ = quicksort(alist)
print(' '.join(str(el) for el in list_))
```
